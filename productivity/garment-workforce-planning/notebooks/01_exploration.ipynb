# Garment workforce productivity exploration
Inspect temporal coverage, departments, staffing, overtime, missing WIP, and target-vs-actual productivity before modeling.

```python
from pathlib import Path
import pandas as pd
import sys
PROJECT = Path('..').resolve()
sys.path.insert(0, str(PROJECT))
from src.data import load_workforce_data
df = load_workforce_data(PROJECT / 'data/raw/garments_worker_productivity.csv')
df.head()
```

```python
df.groupby('department').agg(rows=('team','size'), workers=('no_of_workers','mean'), overtime=('over_time','mean'), actual_productivity=('actual_productivity','mean'))
```

```python
daily = df.groupby('date').agg(target=('targeted_productivity','mean'), actual=('actual_productivity','mean'), workers=('no_of_workers','sum'))
daily[['target','actual']].plot(figsize=(12,4), title='Target vs actual productivity')
```
